# Practice: reading directed acyclic graphs

These cards accompany the [Directed Acyclic Graphs](lecture.ipynb) lecture. Each card shows a graph with a treatment $D$ and an outcome $Y$. Decide which sets of observed variables satisfy the backdoor criterion for the effect of $D$ on $Y$, then open the answer to check your reasoning. Dashed nodes are unobserved and cannot be conditioned on.

The cards grow harder as they go. The first one restates the rules you need for all the others.

## The backdoor criterion

A set of variables $Z$ satisfies the **backdoor criterion** relative to $(D, Y)$ if:

1. No variable in $Z$ is a descendant of $D$
2. $Z$ blocks every backdoor path from $D$ to $Y$

A descendant of $D$ is any node reachable from $D$ by following arrows forward, however many steps away. A backdoor path is any path between $D$ and $Y$ that starts with an arrow into $D$.

### When is a path blocked?

A path is blocked as soon as one node on it blocks it. Whether a node blocks depends on its role on that path and on whether we condition on it:

| Role on the path | Structure | Left alone | Conditioned on |
|------------------|-----------|------------|----------------|
| **Chain** | $D \to M \to Y$ | Lets association through | Blocks |
| **Fork** | $D \leftarrow X \to Y$ | Lets association through | Blocks |
| **Collider** | $D \to C \leftarrow Y$ | Blocks | Lets association through |

Conditioning on a descendant of a collider also lets association through. The same variable can play different roles on different paths.

### A checklist for every card

1. List every path between $D$ and $Y$, ignoring arrow direction
2. Mark the backdoor paths, which start with an arrow into $D$
3. For each backdoor path, note whether it is open or blocked when nothing is conditioned on
4. Find sets that block every backdoor path without opening one through a collider
5. Remove any set that contains a descendant of $D$

![Card 3](practice/card-3.svg)

Which sets of observed variables satisfy the backdoor criterion for the effect of $D$ on $Y$? Bonus: is there more than one valid set?

<details>
<summary>Show answer</summary>

**Nodes**

| Node | Parents | Children | Note |
|------|---------|----------|------|
| $B$ | none | $A$, $Y$ | A confounder: it reaches $D$ through $B \to A \to D$ and causes $Y$ directly |
| $A$ | $B$ | $D$ | Not a confounder, since it reaches $Y$ only through $D$ |
| $D$ | $A$ | $Y$ | Its only descendant is $Y$ |

**Paths**

| Path | Type | Roles | Left alone |
|------|------|-------|------------|
| $D \to Y$ | Causal | none | Open |
| $D \leftarrow A \leftarrow B \to Y$ | Backdoor | $A$ is a chain, $B$ is a fork | Open |

**Valid sets:** {A}, {B}, {A, B}

The backdoor path has no collider, so nothing blocks it until we condition. One blocked node closes the whole path, so either $A$ or $B$ is enough. $A$ is a valid control although it is not a confounder.

</details>

## Card 4

![Card 4](practice/card-4.svg)

Which sets of observed variables satisfy the backdoor criterion for the effect of $D$ on $Y$?

<details>
<summary>Show answer</summary>

**Nodes**

| Node | Parents | Children | Note |
|------|---------|----------|------|
| $A$ | none | $D$, $C$ | A fork on the backdoor path, but not a confounder, since it does not cause $Y$ |
| $B$ | none | $C$, $Y$ | A fork on the backdoor path, but not a confounder, since it does not cause $D$ |
| $C$ | $A$, $B$ | none | A collider |
| $D$ | $A$ | $Y$ | Its only descendant is $Y$ |

**Paths**

| Path | Type | Roles | Left alone |
|------|------|-------|------------|
| $D \to Y$ | Causal | none | Open |
| $D \leftarrow A \to C \leftarrow B \to Y$ | Backdoor | $A$ fork, $C$ collider, $B$ fork | Blocked by $C$ |

**Valid sets:** {}, {A}, {B}, {A, B}, {A, C}, {B, C}, {A, B, C}

This is **M-bias**. The collider $C$ already blocks the backdoor path, so the naive comparison is unbiased. The only invalid set is $\{C\}$, which opens the path. Adding $A$ or $B$ closes it again.

</details>

![Card 5](practice/card-5.svg)

Which sets of observed variables satisfy the backdoor criterion for the effect of $D$ on $Y$?

<details>
<summary>Show answer</summary>

**Nodes**

| Node | Parents | Children | Note |
|------|---------|----------|------|
| $X$ | none | $D$, $Y$ | A confounder |
| $D$ | $X$ | $Y$, $C$ | Its descendants are $Y$ and $C$ |
| $C$ | $D$, $Y$ | none | A collider, and a descendant of $D$ and of $Y$ |

**Paths**

| Path | Type | Roles | Left alone |
|------|------|-------|------------|
| $D \to Y$ | Causal | none | Open |
| $D \leftarrow X \to Y$ | Backdoor | $X$ is a fork | Open |
| $D \to C \leftarrow Y$ | Neither causal nor backdoor | $C$ is a collider | Blocked by $C$ |

**Valid sets:** {X}

The path through $C$ starts with an arrow out of $D$, so the second condition says nothing about it. The first condition rules out $C$ anyway, because $C$ is a descendant of $D$. Conditioning on $C$ would open $D \to C \leftarrow Y$, which is why the criterion excludes descendants of $D$.

</details>

**Nodes**

| Node | Parents | Children | Note |
|------|---------|----------|------|
| $U$ | none | $X$, $Y$ | An unobserved confounder: it reaches $D$ through $U \to X \to D$ |
| $X$ | $U$ | $D$ | Observed, and not a confounder |
| $D$ | $X$ | $Y$ | Its only descendant is $Y$ |

**Paths**

| Path | Type | Roles | Left alone |
|------|------|-------|------------|
| $D \to Y$ | Causal | none | Open |
| $D \leftarrow X \leftarrow U \to Y$ | Backdoor | $X$ is a chain, $U$ is a fork | Open |

**Valid sets:** {X}

We cannot condition on the confounder $U$, but we do not need to. Blocking the backdoor path at the observed chain node $X$ is enough. An unobserved confounder is only fatal when no observed node sits on its backdoor paths.

</details>

## Card 7

![Card 7](practice/card-7.svg)

Which sets of observed variables satisfy the backdoor criterion for the effect of $D$ on $Y$? Bonus: is $C$ a confounder, a collider, or both?

<details>
<summary>Show answer</summary>

**Nodes**

| Node | Parents | Children | Note |
|------|---------|----------|------|
| $A$ | none | $D$, $C$ | A confounder: it causes $D$ directly and $Y$ through $A \to C \to Y$ |
| $B$ | none | $C$, $Y$ | A confounder: it causes $D$ through $B \to C \to D$ and $Y$ directly |
| $C$ | $A$, $B$ | $D$, $Y$ | A confounder, and a collider on $A \to C \leftarrow B$ |
| $D$ | $A$, $C$ | $Y$ | Its only descendant is $Y$ |

**Paths**

| Path | Type | Roles | Left alone |
|------|------|-------|------------|
| $D \to Y$ | Causal | none | Open |
| $D \leftarrow C \to Y$ | Backdoor | $C$ is a fork | Open |
| $D \leftarrow A \to C \to Y$ | Backdoor | $A$ fork, $C$ chain | Open |
| $D \leftarrow C \leftarrow B \to Y$ | Backdoor | $C$ chain, $B$ fork | Open |
| $D \leftarrow A \to C \leftarrow B \to Y$ | Backdoor | $A$ fork, $C$ collider, $B$ fork | Blocked by $C$ |

**Valid sets:** {A, C}, {B, C}, {A, B, C}

$C$ is both. The path $D \leftarrow C \to Y$ can only be blocked at $C$, but conditioning on $C$ opens the last path, so we also need $A$ or $B$ to close it. Neither "control for every confounder" ($\{C\}$) nor "never control for a collider" ($\{A, B\}$) gives a valid set. Only checking every path does.

</details>

## Card 8

![Card 8](practice/card-8.svg)

Which sets of observed variables satisfy the backdoor criterion for the effect of $D$ on $Y$? $U$ is unobserved. Answer for the total effect of $D$ on $Y$, then think about the direct effect that does not run through $M$.

<details>
<summary>Show answer</summary>

**Nodes**

| Node | Parents | Children | Note |
|------|---------|----------|------|
| $D$ | none | $M$, $Y$ | Its descendants are $M$ and $Y$ |
| $M$ | $D$, $U$ | $Y$ | A mediator, and a collider between $D$ and $U$ |
| $U$ | none | $M$, $Y$ | Unobserved |

**Paths**

| Path | Type | Roles | Left alone |
|------|------|-------|------------|
| $D \to Y$ | Causal | none | Open |
| $D \to M \to Y$ | Causal | $M$ is a chain | Open |
| $D \to M \leftarrow U \to Y$ | Neither causal nor backdoor | $M$ collider, $U$ fork | Blocked by $M$ |

**Valid sets:** {}

$D$ has no parents, so there is no backdoor path and the naive comparison identifies the total effect. The direct effect is harder. Conditioning on $M$ blocks the mediated path as intended, but it also opens $D \to M \leftarrow U \to Y$. Because $U$ is unobserved, nothing closes that path again, and no conditioning set recovers the direct effect. This is the structure of the discrimination example in the lecture.

</details>
